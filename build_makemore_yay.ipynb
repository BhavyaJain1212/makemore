```python
N = torch.zeros((27, 27), dtype=torch.int32)
```

```python
chars = sorted(list(set((''.join(words)))))
stoi = {}
for i, s in enumerate(chars):
    stoi[s] = i+1

stoi['.'] = 0

stoi
```

```python
itos = {}
for key, val in stoi.items():
    itos[val] = key

itos
```

```python
for w in words:
    chs = ['.'] + list(w) + ['.'] # adding start and end token to each word
    # print(list(zip(chs, chs[1:])))
    for ch1, ch2 in zip(chs, chs[1:]):
        # [('<S>', 'e'), ('e', 'm'), ('m', 'm'), ('m', 'a'), ('a', '<E>')]
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
```

```python
log_likelihood = 0.0
n = 0
for w in words:
    chs = ['.'] + list(w) + ['.'] # adding start and end token to each word
    # print(list(zip(chs, chs[1:])))
    for ch1, ch2 in zip(chs, chs[1:]):
        # [('<S>', 'e'), ('e', 'm'), ('m', 'm'), ('m', 'a'), ('a', '<E>')]
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        prob = P[ix1, ix2]
        logprob = torch.log(prob)
        log_likelihood += logprob
        n += 1
        # print(f'{ch1}{ch2}: {prob:.4f} {logprob:.4f}')

print(f'{log_likelihood=}')
nll = -log_likelihood
print(f'{nll=}')
print(f'{nll/n}')
```

```python
# creating the dataset
xs, ys = [], []

for w in words:
    chs = ['.'] + list(w) + ['.'] # adding start and end token to each word
    # print(list(zip(chs, chs[1:])))
    for ch1, ch2 in zip(chs, chs[1:]):
        # [('<S>', 'e'), ('e', 'm'), ('m', 'm'), ('m', 'a'), ('a', '<E>')]
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
        xs.append(ix1)
        ys.append(ix2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)
num = xs.nelement() # number of elements
print('number of examples: ', num)

# initializing the network
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27, 27), generator=g, requires_grad=True)
```

```python
# gradient descent

for k in range(100):

    # forward pass
    xenc = F.one_hot(xs, num_classes=27).float()
    logits = (xenc @ W) # predict log-counts
    counts = logits.exp() # equivalent to N matrix
    probs = counts / counts.sum(1, keepdim=True) # the above two lines are what we call a softmax
    loss = -probs[torch.arange(num), ys].log().mean() # + 0.1*(W**2).mean() # regularization
    print(loss.item())

    # backward pass
    W.grad = None # set gradients to zero
    loss.backward()

    # update
    W.data += -50 * W.grad
```

```python
g = torch.Generator().manual_seed(2147483647)

for i in range(5):
    out = []
    ix = 0
    while True:
        xenc = F.one_hot(torch.tensor([ix]), num_classes=27).float()
        logits = xenc @ W # predict log-counts
        counts = logits.exp() # equivalent to N matrix
        p = counts / counts.sum(1, keepdim=True)

        ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
        out.append(itos[ix])
        if ix == 0:
            break
    print(''.join(out))
```
